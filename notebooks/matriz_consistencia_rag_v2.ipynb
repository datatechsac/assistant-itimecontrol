## 1. Situación problemática / necesidad real

El sistema **iTimeControl** es una plataforma de control de asistencia y gestión del tiempo utilizada por personal operativo (administradores de RRHH, supervisores y empleados) para registrar marcaciones, gestionar horarios, generar reportes y administrar permisos.

En la práctica, estos usuarios enfrentan una **necesidad real**: cuando tienen una duda operativa (por ejemplo, *“¿cómo corrijo una marcación?”* o *“¿cómo genero el reporte de tardanzas?”*), deben recurrir a manuales extensos o esperar la respuesta del soporte técnico. Esto genera:

- Demoras operativas y dependencia del área de soporte.
- Respuestas inconsistentes según quién atienda la consulta.
- Una curva de aprendizaje elevada para usuarios nuevos.
- Sobrecarga del personal de soporte con preguntas repetitivas.

La documentación existe, pero está dispersa y no es consultable en lenguaje natural. **Falta un medio que entregue respuestas inmediatas, precisas y fundamentadas en la documentación oficial del sistema.**

## 9. Avances siguiente semana — Análisis de slices problemáticos

Como parte de la mejora continua del sprint, se analiza el desempeño del asistente **por segmentos (slices)** de consultas, en lugar de mirar solo la métrica global. Un promedio alto puede ocultar categorías donde el sistema falla.

Para cada slice se reporta:

- **Métrica** (ROUGE-L media de recuperación).
- **Intervalo de confianza (IC 95%)** por bootstrap — indica cuán confiable es la media.
- **Tamaño (N)** — número de consultas en el slice.

Luego se identifican los slices problemáticos, su **causa probable con evidencia** y un **plan de mitigación**.


```python
import json, warnings
import numpy as np
import pandas as pd
from pathlib import Path
from collections import defaultdict
warnings.filterwarnings('ignore')
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from rouge_score import rouge_scorer

np.random.seed(42)

# Cargar dataset (con campo 'cat' = slice/categoria de la consulta)
candidates = [
    Path.cwd() / "data" / "datasets",
    Path.cwd().parent / "data" / "datasets",
    Path("../data/datasets"),
    Path("../../itimecontrol-assistant/data/datasets"),
    Path("../../data/datasets"),
]

D = next((p for p in candidates if p.exists()), None)

if D is None:
    raise FileNotFoundError(
        "No se encontró la carpeta de datasets. Revisá una de estas rutas: "
        + ", ".join(str(p) for p in candidates)
    )


def load(p):
    return [json.loads(l) for l in Path(p).read_text(encoding='utf-8').splitlines() if l.strip()]

try:
    alld = load(D / "train.jsonl") + load(D / "val.jsonl") + load(D / "test.jsonl")
    print(f"✅ Datasets cargados desde: {D}")
except FileNotFoundError:
    print(f"⚠️  No se encontraron los archivos train/val/test en {D}.")
    alld = []

# Deduplicar por pregunta
seen=set(); data=[]
for r in alld:
    if r['instruction'] not in seen:
        seen.add(r['instruction']); data.append(r)
print(f'Consultas únicas: {len(data)}')
if data:
    cats = sorted(set(r.get('cat','general') for r in data))
    print(f'Slices (categorías): {cats}')
```

```python
# Construir retriever TF-IDF y evaluar por muestra (leave-one-out)
if data:
    corpus = [r['output'] for r in data]
    vec = TfidfVectorizer(ngram_range=(1,2), max_features=2000, strip_accents='unicode')
    M = vec.fit_transform(corpus)
    scorer = rouge_scorer.RougeScorer(['rougeL'], use_stemmer=False)

    per_sample = []
    for i, r in enumerate(data):
        q, ref, cat = r['instruction'], r['output'], r.get('cat','general')
        v = vec.transform([q])
        sims = cosine_similarity(v, M).flatten()
        sims[i] = -1  # excluir el propio documento (evita leakage trivial)
        pred = corpus[sims.argmax()]
        rl = scorer.score(ref, pred)['rougeL'].fmeasure
        per_sample.append({'slice': cat, 'rougeL': rl})

    print(f'Muestras evaluadas: {len(per_sample)}')
    print('✅ Evaluación por muestra completada')
```

```python
# Agregación por slice: métrica + IC 95% (bootstrap) + tamaño N
def bootstrap_ci(vals, n_boot=1000, alpha=0.05):
    """IC de la media por bootstrap. Con N=1 devuelve el propio valor."""
    vals = np.asarray(vals, dtype=float)
    if len(vals) < 2:
        return (float(vals[0]), float(vals[0]))
    boot = [np.mean(np.random.choice(vals, len(vals), replace=True)) for _ in range(n_boot)]
    lo = np.percentile(boot, 100*alpha/2)
    hi = np.percentile(boot, 100*(1-alpha/2))
    return (float(lo), float(hi))

if data:
    by_slice = defaultdict(list)
    for s in per_sample:
        by_slice[s['slice']].append(s['rougeL'])

    global_mean = float(np.mean([s['rougeL'] for s in per_sample]))
    UMBRAL = 0.85  # un slice es problemático si su media < 85% de la media global

    rows = []
    for cat, vals in by_slice.items():
        m = float(np.mean(vals)); n = len(vals)
        lo, hi = bootstrap_ci(vals)
        problematico = m < global_mean * UMBRAL
        rows.append({
            'Slice': cat, 'N': n, 'ROUGE-L': round(m,3),
            'IC 95%': f'[{lo:.3f}, {hi:.3f}]',
            'Estado': '🔴 PROBLEMÁTICO' if problematico else '🟢 OK'
        })

    df_slices = pd.DataFrame(rows).sort_values('ROUGE-L').reset_index(drop=True)
    print(f'Media global ROUGE-L: {global_mean:.3f}  |  Umbral problemático: < {global_mean*UMBRAL:.3f}\n')
    display(df_slices)

    problematicos = df_slices[df_slices['Estado'].str.contains('PROBLEMÁTICO')]
    print(f'\nSlices problemáticos detectados: {len(problematicos)} de {len(df_slices)}')
```

### 9.1. Diagnóstico de slices problemáticos

Para cada slice problemático detectado se documenta la **causa probable (con evidencia)** y su **plan de mitigación**. La siguiente tabla sirve como plantilla; ajusta las causas según los slices que arroje tu ejecución.

| Slice problemático | Métrica (ROUGE-L) + IC + N | Causa probable (con evidencia) | Plan de mitigación |
|--------------------|----------------------------|--------------------------------|--------------------|
| **permisos** | Baja media, IC estrecho, N pequeño | La documentación de permisos/solicitudes tiene menos fragmentos en el corpus; el vocabulario de la consulta ("justificar", "vacaciones") no coincide con el de los documentos. Evidencia: nº de chunks de esta categoría es de los más bajos. | Ampliar la documentación de permisos; añadir sinónimos/expansión de consulta; aumentar el peso de esta categoría en la indexación. |
| **configuración** | Media baja con IC amplio | Categoría muy heterogénea (backup, roles, dispositivos, contraseñas) agrupada en un solo slice; el retriever confunde subtemas. Evidencia: IC 95% ancho indica alta varianza interna. | Dividir "configuración" en subcategorías más finas; crear chunks más específicos por subtema. |
| **horas_extras / empleados** | N muy pequeño (N=1) | Muestra insuficiente para conclusiones fiables. Evidencia: IC colapsa a un punto (N=1). | Recolectar más consultas reales de estas categorías antes de sacar conclusiones; no priorizar mitigación hasta tener N≥5. |

> **Criterio de interpretación:** un slice con **N pequeño** y **IC muy ancho** no es concluyente — puede parecer problemático solo por azar. Se prioriza mitigar los slices con **N suficiente (≥5)** y media consistentemente baja.


```python
# Generar automáticamente el plan de mitigación priorizado según N e IC
if data:
    plan = []
    for _, r in df_slices.iterrows():
        if 'PROBLEMÁTICO' not in r['Estado']:
            continue
        n = r['N']
        if n >= 5:
            prioridad = 'ALTA'
            accion = 'Mitigar ya: ampliar documentación y mejorar chunks de este slice.'
        elif n >= 2:
            prioridad = 'MEDIA'
            accion = 'Revisar vocabulario/sinónimos; recolectar más ejemplos.'
        else:
            prioridad = 'BAJA (no concluyente)'
            accion = 'Recolectar más datos antes de decidir (N=1 no es fiable).'
        plan.append({'Slice': r['Slice'], 'N': n, 'ROUGE-L': r['ROUGE-L'],
                     'Prioridad': prioridad, 'Acción recomendada': accion})

    if plan:
        df_plan = pd.DataFrame(plan)
        print('PLAN DE MITIGACIÓN PRIORIZADO:')
        display(df_plan)
        df_plan.to_csv('plan_mitigacion_slices.csv', index=False, encoding='utf-8-sig')
        print('\n✅ Plan exportado a: plan_mitigacion_slices.csv')
    else:
        print('✅ No se detectaron slices problemáticos con los datos actuales.')
```
