## 1. Situación problemática / necesidad real

El sistema **iTimeControl** es una plataforma de control de asistencia y gestión del tiempo utilizada por personal operativo (administradores de RRHH, supervisores y empleados) para registrar marcaciones, gestionar horarios, generar reportes y administrar permisos.

En la práctica, estos usuarios enfrentan una **necesidad real**: cuando tienen una duda operativa (por ejemplo, *“¿cómo corrijo una marcación?”* o *“¿cómo genero el reporte de tardanzas?”*), deben recurrir a manuales extensos o esperar la respuesta del soporte técnico. Esto genera:

- Demoras operativas y dependencia del área de soporte.
- Respuestas inconsistentes según quién atienda la consulta.
- Una curva de aprendizaje elevada para usuarios nuevos.
- Sobrecarga del personal de soporte con preguntas repetitivas.

La documentación existe, pero está dispersa y no es consultable en lenguaje natural. **Falta un medio que entregue respuestas inmediatas, precisas y fundamentadas en la documentación oficial del sistema.**

---
## 2. Problema de investigación

### Problema general

> ¿En qué medida la implementación de un asistente conversacional inteligente basado en RAG mejora la resolución de consultas operativas en el sistema iTimeControl?

### Problemas específicos

| Nº | Problema específico |
|----|---------------------|
| PE1 | ¿Cómo construir una base de conocimiento consultable a partir de la documentación de iTimeControl? |
| PE2 | ¿Cómo influye la calidad de la recuperación semántica (retriever) en la precisión de las respuestas a consultas operativas? |
| PE3 | ¿Cómo generar respuestas en lenguaje natural fundamentadas en el contexto recuperado? |
| PE4 | ¿En qué medida el asistente basado en RAG supera a un método de recuperación tradicional (baseline)? |

---
## 3. Objetivos

### Objetivo general

> Implementar un asistente conversacional inteligente basado en RAG para mejorar la resolución de consultas operativas en el sistema iTimeControl.

### Objetivos específicos

| Nº | Objetivo específico |
|----|---------------------|
| OE1 | Construir una base de conocimiento vectorizada a partir de la documentación oficial de iTimeControl mediante extracción, limpieza y normalización de chunks. |
| OE2 | Implementar el pipeline de recuperación semántica (embeddings + base vectorial + retriever) que seleccione los fragmentos más relevantes ante una consulta. |
| OE3 | Integrar el componente de generación de respuestas que use el contexto recuperado para responder en lenguaje natural. |
| OE4 | Evaluar el desempeño del asistente RAG mediante métricas objetivas (Recall@k, ROUGE, BLEU) comparándolo con un baseline de recuperación tradicional. |

---
## 4. Hipótesis

### Hipótesis general (H1)

> La implementación de un asistente conversacional basado en RAG mejora significativamente la resolución de consultas operativas en el sistema iTimeControl, en comparación con un método de recuperación tradicional.

### Hipótesis nula (H0)

> El asistente basado en RAG no presenta diferencias significativas en la resolución de consultas operativas frente a un método de recuperación tradicional.

### Hipótesis específicas

| Nº | Hipótesis específica |
|----|----------------------|
| He1 | La normalización de chunks mejora la homogeneidad y relevancia de los fragmentos frente a un chunking sin normalizar. |
| He2 | A mayor calidad del retriever (recuperación semántica), mayor es la precisión de las respuestas. |
| He3 | Las respuestas generadas con contexto recuperado son más precisas y completas que sin contexto. |
| He4 | El pipeline RAG completo obtiene métricas de calidad (ROUGE, BLEU, Recall@k) superiores al baseline. |

## 5. Variables

Las variables se derivan del **objetivo general** (y por tanto del título completo de la tesis), no de cada objetivo específico. Los objetivos específicos son las **etapas** para alcanzar el objetivo general.

| Tipo | Variable | Definición conceptual |
|------|----------|----------------------|
| **Independiente (VI)** | Asistente conversacional basado en RAG | Sistema que combina recuperación semántica de documentos (embeddings + base vectorial) con generación de respuestas en lenguaje natural. Incluye la configuración del chunking, el top-k de recuperación y la generación aumentada por contexto. |
| **Dependiente (VD)** | Resolución de consultas operativas | Grado en que el sistema responde de forma correcta, relevante y completa las consultas operativas sobre el sistema iTimeControl. |
| **Control** | Condiciones constantes del experimento | Modelo de embeddings utilizado, corpus documental, semilla aleatoria, partición train/val/test y conjunto de preguntas de evaluación. |

## 6. Operacionalización de variables

Cada variable se descompone en dimensiones medibles con indicadores anclados en componentes reales del proyecto.

### 6.1. Variable Independiente: Asistente conversacional basado en RAG

| Dimensión | Indicadores | Instrumento / Técnica |
|-----------|-------------|----------------------|
| Procesamiento documental | • Coeficiente de variación del tamaño de chunks<br>• Número de chunks generados<br>• Dimensión del vector de embeddings | Script de chunking normalizado; reporte de distribución de tamaños |
| Recuperación semántica | • Valor de top-k configurado<br>• Umbral de similitud coseno<br>• Tipo de base vectorial (FAISS/ChromaDB) | Pipeline retriever; registro de configuración |
| Generación aumentada | • Longitud máxima de contexto inyectado<br>• Número de fragmentos usados por respuesta | Módulo de generación; logs del pipeline |

### 6.2. Variable Dependiente: Resolución de consultas operativas

| Dimensión | Indicadores | Instrumento / Técnica |
|-----------|-------------|----------------------|
| Relevancia de la recuperación | • Recall@1<br>• Recall@3<br>• Recall@5<br>• MRR (Mean Reciprocal Rank) | Script de evaluación sobre conjunto de test |
| Precisión de la respuesta | • ROUGE-1<br>• ROUGE-2<br>• ROUGE-L | Métricas automáticas (rouge-score) |
| Fidelidad lingüística | • BLEU | Métrica automática (sacrebleu / nltk) |
| Eficiencia operativa | • Latencia de respuesta (ms) | Medición de tiempo en el pipeline |

## 8. Verificación de coherencia — ¿La matriz "cierra"?

La siguiente celda valida programáticamente que cada eslabón de la cadena metodológica tenga su correspondencia, es decir, que ningún problema, objetivo, hipótesis o indicador quede "huérfano".

```python
import pandas as pd

# Definición de la matriz como estructura de datos
matriz = [
    {
        'codigo': 'E1',
        'problema':  'PE1: construir base de conocimiento consultable',
        'objetivo':  'OE1: construir base vectorizada (chunking normalizado)',
        'hipotesis': 'He1: la normalización mejora homogeneidad y relevancia',
        'variable':  'VI / Procesamiento documental',
        'indicador': 'Coef. variación tamaño chunks; nº chunks',
        'instrumento':'Script chunking; reporte distribución',
    },
    {
        'codigo': 'E2',
        'problema':  'PE2: influencia de la recuperación en la precisión',
        'objetivo':  'OE2: implementar retriever semántico',
        'hipotesis': 'He2: mejor retriever -> mayor precisión',
        'variable':  'VI / Recuperación · VD / Relevancia',
        'indicador': 'Top-k; umbral similitud; Recall@1,3,5',
        'instrumento':'Pipeline FAISS/Chroma; script evaluación',
    },
    {
        'codigo': 'E3',
        'problema':  'PE3: generar respuestas fundamentadas en contexto',
        'objetivo':  'OE3: integrar generación con contexto recuperado',
        'hipotesis': 'He3: con contexto -> respuestas más precisas',
        'variable':  'VI / Generación · VD / Precisión',
        'indicador': 'Nº fragmentos; ROUGE-1/2/L; BLEU',
        'instrumento':'Módulo generación; métricas automáticas',
    },
    {
        'codigo': 'E4',
        'problema':  'PE4: superioridad del RAG frente al baseline',
        'objetivo':  'OE4: evaluar RAG vs baseline',
        'hipotesis': 'He4: RAG supera al baseline',
        'variable':  'VD / Eficiencia y comparación',
        'indicador': 'Diferencia ROUGE/BLEU/Recall@k; latencia',
        'instrumento':'Experimento comparativo; holdout + 5-fold CV',
    },
]

df = pd.DataFrame(matriz).set_index('codigo')

# Verificar que ninguna celda esté vacía (todos los eslabones presentes)
print('='*60)
print('  VERIFICACIÓN DE COHERENCIA DE LA MATRIZ')
print('='*60)
columnas_clave = ['problema', 'objetivo', 'hipotesis', 'variable', 'indicador', 'instrumento']
todo_completo = True
for cod, fila in df.iterrows():
    faltantes = [c for c in columnas_clave if not str(fila[c]).strip()]
    estado = '✅ CIERRA' if not faltantes else f'❌ falta: {faltantes}'
    print(f'  Eslabón {cod}: {estado}')
    if faltantes:
        todo_completo = False

print('='*60)
if todo_completo:
    print('  ✅ LA MATRIZ CIERRA: cada eslabón tiene correspondencia completa')
    print('     (problema → objetivo → hipótesis → variable → indicador → instrumento)')
else:
    print('  ❌ La matriz NO cierra: revisar los eslabones marcados')
print('='*60)
```

```python
# Mostrar la matriz completa como tabla
pd.set_option('display.max_colwidth', None)
pd.set_option('display.width', None)

df_display = df.rename(columns={
    'problema': 'Problema Específico',
    'objetivo': 'Objetivo Específico',
    'hipotesis': 'Hipótesis Específica',
    'variable': 'Variable / Dimensión',
    'indicador': 'Indicador',
    'instrumento': 'Instrumento',
})
df_display
```

```python
# Exportar la matriz a CSV (útil para anexar al documento de tesis)
df_display.to_csv('matriz_consistencia_rag_itimecontrol.csv', encoding='utf-8-sig')
print('✅ Matriz exportada a: matriz_consistencia_rag_itimecontrol.csv')

# Resumen final
print('\n' + '='*60)
print('  RESUMEN DE LA MATRIZ DE CONSISTENCIA')
print('='*60)
print(f'  Título: Asistente conversacional basado en RAG para iTimeControl')
print(f'  Eslabones (objetivos específicos): {len(df)}')
print(f'  Variable independiente: Asistente RAG')
print(f'  Variable dependiente: Resolución de consultas operativas')
print(f'  Indicadores clave: Recall@k, ROUGE, BLEU, latencia')
print('='*60)
```

### Ventajas metodológicas de elegir RAG

1. **Menor costo computacional**: no requiere GPU para entrenar un modelo desde cero.
2. **Actualización sencilla**: si la documentación de iTimeControl cambia, basta con re-indexar.
3. **Trazabilidad**: cada respuesta puede citar el fragmento de origen, reduciendo alucinaciones.
